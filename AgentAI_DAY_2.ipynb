```python
from google.colab import userdata

groq_api_key = userdata.get("GROQ_API_KEY")

print("API Key Loaded:", bool(groq_api_key))
```

```python
from groq import Groq

client = Groq(api_key=groq_api_key)

MODEL = "openai/gpt-oss-120b"

print("Groq client initialized successfully!")
```

```python
messages = [
    {
        "role": "system",
        "content": "You are a helpful travel information assistant."
    },
    {
        "role": "user",
        "content": "Suggest 3 tourist places in Hyderabad."
    }
]
response = client.chat.completions.create(
    model=MODEL,
    messages=messages
)
print(response.choices[0].message.content)
```

```python
system_prompt = """
You are a Smart Travel Information Assistant.

Your responsibilities:
1. Suggest tourist places.
2. Provide travel tips.
3. Create simple travel plans.
4. Ask questions when information is missing.
5. Never invent live train or bus availability.
6. Give answers in simple English.
"""

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": "Plan a 2-day trip to Hyderabad."}
]

response = client.chat.completions.create(
    model=MODEL,
    messages=messages
)

print(response.choices[0].message.content)
```

```python
few_shot_prompt = """
You are a travel assistant.

Example 1:
User: I want to visit Hyderabad.
Assistant: Hyderabad has Charminar, Golconda Fort and Hussain Sagar.

Example 2:
User: I want to visit Delhi.
Assistant: Delhi has India Gate, Red Fort and Qutub Minar.

Now answer the next question in the same style.
"""

messages = [
    {"role": "system", "content": few_shot_prompt},
    {"role": "user", "content": "I want to visit Chennai."}
]

response = client.chat.completions.create(
    model=MODEL,
    messages=messages
)

print(response.choices[0].message.content)
```

```python
import json

json_prompt = """
Suggest a 2-day trip to Hyderabad.

Return only valid JSON in this format:
{
    "destination": "city name",
    "duration": "number of days",
    "places": ["place 1", "place 2"],
    "tips": ["tip 1", "tip 2"]
}
"""

messages = [
    {"role": "system", "content": "You are a travel assistant. Follow the requested JSON format."},
    {"role": "user", "content": json_prompt}
]

response = client.chat.completions.create(
    model=MODEL,
    messages=messages
)

result = response.choices[0].message.content

print(result)
```

```python
try:
    travel_data = json.loads(result)

    print("Destination:", travel_data["destination"])
    print("Duration:", travel_data["duration"])
    print("Places:", travel_data["places"])
    print("Tips:", travel_data["tips"])

except json.JSONDecodeError:
    print("Invalid JSON response. Please try again.")
```

```python
def travel_assistant(destination):
    prompt = f"""
    Suggest 5 tourist places in {destination}.
    Give a short explanation for each place.
    """
    messages = [
        {
            "role": "system",
            "content": "You are a helpful travel assistant."
        },
        {
            "role": "user",
            "content": prompt
        }
    ]
    response = client.chat.completions.create(
        model=MODEL,
        messages=messages
    )
    return response.choices[0].message.content
answer = travel_assistant("Hyderabad")
print(answer)
```

```python
def show_message(func):
    def wrapper(*args, **kwargs):
        print("Welcome to Smart Travel Assistant!")
        result = func(*args, **kwargs)
        print("Thank you for using Travel Assistant!")
        return result
    return wrapper
@show_message
def greet_user():
    print("Let's plan your next trip!")
greet_user()
```

```python
import json

memory = {}

def save_details(destination, duration, budget):
    memory["destination"] = destination
    memory["duration"] = duration
    memory["budget"] = budget

def generate_travel_plan(question):
    context = f"""
    User Travel Details:
    Destination: {memory.get('destination', 'Not provided')}
    Duration: {memory.get('duration', 'Not provided')} days
    Budget: {memory.get('budget', 'Not provided')}
    """
    messages = [
        {
            "role": "system",
            "content": """
            You are a Smart Travel Information Assistant.
            Give simple and useful travel suggestions.
            Ask for missing information.
            Never invent live train or bus availability.
            """
        },
        {
            "role": "user",
            "content": context + "\nQuestion: " + question
        }
    ]

    response = client.chat.completions.create(
        model=MODEL,
        messages=messages
    )
    return response.choices[0].message.content

try:
    destination = input("Enter destination: ")
    duration = int(input("Enter number of days: "))
    budget = int(input("Enter your budget: "))
    if not destination.strip():
        raise ValueError("Destination cannot be empty.")
    if duration <= 0 or budget <= 0:
        raise ValueError("Duration and budget must be positive.")
    save_details(destination, duration, budget)
    print("\nYour travel details:")
    print(memory)
    question = input("\nWhat do you want to know about your trip? ")
    print("\nAI Travel Assistant Response:\n")
    answer = generate_travel_plan(question)
    print(answer)
except ValueError as error:
    print("Input Error:", error)
except Exception as error:
    print("Error:", error)
```
