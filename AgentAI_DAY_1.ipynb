```python
from google.colab import userdata

groq_api_key = userdata.get("GROQ_API_KEY")

print("API key loaded:", bool(groq_api_key))
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    api_key=groq_api_key,
    temperature=0
)
```

```python
SYSTEM = """
You are a Travel Information Assistant.

Your responsibilities:
1. Help users plan trips and understand travel destinations.
2. If the destination is missing, ask the user which place they want to visit.
3. Suggest tourist places, activities and simple travel itineraries.
4. Never invent live train, bus, flight availability or ticket prices.
5. If the user asks about live availability, clearly explain that
   live availability cannot be verified.
6. Use simple English.
7. Give responses in the following format:

Destination:
Travel Information:
Suggested Plan:
Important Note:

Keep the response simple, clear and helpful.
"""
```

```python
def ask(question):
    response = llm.invoke([
        ("system", SYSTEM),
        ("human", question)
    ])

    return response.content
```

```python
print(ask("I want to travel tomorrow"))
```

```python
print(ask("Suggest places to visit in Hyderabad"))
```

```python
print(ask("Is a train available tomorrow at 8 AM?"))
```

```python
print(ask("Plan a 2-day trip"))
```

```python
print(ask("Plan a 2-day trip to Hyderabad"))
```
